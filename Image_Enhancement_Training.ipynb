## Train 1- CLAHE

````python
import yaml
import os
from ultralytics import YOLO

# 1. Define paths for CLAHE
base_dir = "/content/drive/MyDrive/Thesis_Pothole_Project"
ds_path = f"{base_dir}/Dataset_2_CLAHE"
algo_name = "CLAHE"

print(f"INITIALIZING OPTIMIZED TRAINING FOR: {algo_name}")

# 2. Dynamically generate the data.yaml for CLAHE
original_yaml_path = f"{base_dir}/Dataset_1_Baseline/data.yaml"
algo_yaml_path = os.path.join(ds_path, f"data_{algo_name.lower()}.yaml")

with open(original_yaml_path, 'r') as f:
    base_config = yaml.safe_load(f)

base_config['train'] = os.path.join(ds_path, 'train', 'images')
base_config['val'] = os.path.join(ds_path, 'valid', 'images')
if 'test' in base_config:
    del base_config['test']

with open(algo_yaml_path, 'w') as f:
    yaml.dump(base_config, f)

# 3. Load Model and Train with Strict Hyperparameters
model = YOLO("yolov8n.pt")

model.train(
    data=algo_yaml_path,
    epochs=150,
    patience=15,                 # Safely stops if accuracy peaks early
    imgsz=640,
    batch=16,
    project=f"{base_dir}/Training_Results",
    name=f"Optimized_Model_{algo_name}_YOLOv8n",
    exist_ok=True,

    hsv_h=0.0,                   # Disable Hue shifting
    hsv_s=0.0,                   # Disable Saturation shifting
    hsv_v=0.0,                   # Disable Brightness/Value shifting
    mosaic=0.0                   # Disable image scrambling
)
print(f"Optimized Training finished for {algo_name}!")
````

```markdown
## Summary and Comparison of Results
This section compiles and compares the performance metrics (`Precision`, `Recall`, `mAP50`, `mAP50-95`) of all the trained YOLOv8 models from the GDrive directory.
```

````python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

training_results_dir = "/content/drive/MyDrive/Thesis_Pothole_Project/Training_Results"
results_data = []

if os.path.exists(training_results_dir):
    for model_dir in os.listdir(training_results_dir):
        model_path = os.path.join(training_results_dir, model_dir)
        results_csv = os.path.join(model_path, "results.csv")

        if os.path.isdir(model_path) and os.path.exists(results_csv):
            try:
                df_run = pd.read_csv(results_csv)
                df_run.columns = df_run.columns.str.strip()

                # Get metrics from the best epoch or final validation epoch
                best_epoch = df_run.iloc[-1]

                results_data.append({
                    "Model Name": model_dir,
                    "mAP50": best_epoch.get("metrics/mAP50(B)", None),
                    "mAP50-95": best_epoch.get("metrics/mAP50-95(B)", None),
                    "Precision": best_epoch.get("metrics/precision(B)", None),
                    "Recall": best_epoch.get("metrics/recall(B)", None),
                })
            except Exception as e:
                print(f"Could not parse results for {model_dir}: {e}")

    df_results = pd.DataFrame(results_data)
    if not df_results.empty:
        # Sort models by mAP50 for clearer comparison
        df_results = df_results.sort_values(by="mAP50", ascending=False).reset_index(drop=True)
        display(df_results)
    else:
        print("No valid results.csv files found.")
else:
    print(f"Directory {training_results_dir} does not exist.")
````

````python
if 'df_results' in locals() and not df_results.empty:
    plt.figure(figsize=(14, 8))

    # Melt the dataframe for seaborn barplot visualization
    df_melted = df_results.melt(id_vars="Model Name",
                                 value_vars=["mAP50", "mAP50-95", "Precision", "Recall"],
                                 var_name="Metric", value_name="Value")

    sns.barplot(data=df_melted, x="Value", y="Model Name", hue="Metric", palette="viridis")
    plt.title("Comparison of YOLOv8 Models across Evaluation Metrics")
    plt.xlabel("Score")
    plt.ylabel("Model Configuration")
    plt.xlim(0.0, 1.0)
    plt.grid(True, linestyle="--", alpha=0.6)
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
````
